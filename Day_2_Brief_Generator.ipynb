```python
def save_brief(brief, filename="brief.md"):
    lines = ["# Problem Brief", ""]
    for key, value in brief.items():
        lines.append("## " + key.upper())
        lines.append(value)
        lines.append("")
    with open(filename, "w") as f:
        f.write("\n".join(lines))
    print("Saved to", filename)

brief = make_brief(
    "Our support team drowns in refund emails.",
    "Support team overloaded by refund emails",
    "The support team",
    "Emails sorted automatically, reply drafts ready to review"
)
save_brief(brief)
```
